```python
import random
import numpy as np
import math
```

#Part 1

```python
#Last two digit of my id is 41

R = 41
radius = R / 10.0

full_name = "Abdullah Toufiq"
num_letters = len(full_name.replace(" ", ""))
base_points = 200 * num_letters

print("Radius: ", radius)
print("Number of letters: ", num_letters)
print("Base Point: ", base_points)
```

#Part 2

```python
def estimate_pi(total_points, radius):
  inside = 0
  half = total_points // 2

  for _ in range(half):
    x = random.uniform(0, radius)
    y = random.uniform(0, radius)
    if x*x + y*y <= radius*radius:
      inside += 1


  #other half
  xo = np.random.rand(half)
  yo = np.random.rand(half)
  inside += np.sum(xo*xo + yo*yo <= radius*radius)

  estimated_pi = 4.0 * inside / total_points
  return estimated_pi, inside
```

#Part 3

```python
multipliers = [1,2,3,4,5]

print("|Ranged used|Total points|Points inside|Est pi    |Error from pi  ")

for i in multipliers:
  total = base_points * i
  est_pi, inside = estimate_pi(total, radius)

  error = est_pi - math.pi
  print(f"|[0, {radius}]   |{total}        |{inside}         |{est_pi:.5f}   |{error}")
```
