# Uniform Distribution with Random Numbers

A uniform distribution is a type of probability distribution where all outcomes are equally likely. In other words, every event in a given range has the same chance of happening. For example, when you roll a fair six-sided die, the chances of rolling any of the numbers 1 through 6 are all the same, making the distribution of outcomes uniform.

- **Creates a histogram** of the values in `random_numbers`.  
- **`bins=50`** → Divides the range (0 to 1) into 50 equal-width intervals.
- **`alpha=0.75`** → Sets transparency; `1.0` is fully opaque, `0.0` is fully transparent.
- **`color='blue'`** → Fills histogram bars with blue.  
- **`edgecolor='black'`** → Outlines each bin in black for better visibility.  


#Normal Distribution

A **normal distribution**, also known as a **Gaussian distribution**, is a type of probability distribution that is symmetrical and bell-shaped. In this distribution:

- Most of the data points are clustered around the **mean** (average).
- As you move away from the mean, the probability of the data points decreases, but in a smooth, continuous way.
- The curve is symmetrical, meaning the left side is a mirror image of the right side.

In simpler terms, a normal distribution looks like a bell curve, where most values are near the middle, and fewer values are found as you move further away from the center. Examples include things like people's heights, test scores, or measurement errors.

## Part 1 - Date generation

```python
# Generate 3000 random numbers between 2 to 12 minutes
uniform_distributionValues = np.random.uniform(2,12,3000)

# Generate 3000 random values from a normal distribution
NormalValues = np.random.normal(8, 2, 3000) #mean=8, standard deviation = 2

#Replacing negative value with 0
NormalValues[NormalValues < 0] = 0

```

#Part 2 - Visualization

```python
# Create both histograms in the same figure
plt.figure(figsize=(15, 5))

# Histogram for Uniform Distribution
plt.subplot(1, 2, 1)
plt.hist(uniform_distributionValues, bins=30, edgecolor='black')
plt.title('Histogram of Uniform Distribution between 2 and 12 minutes')
plt.xlabel('Waiting Time (minutes)')
plt.ylabel('Frequency')

# Histogram for Normal Distribution
plt.subplot(1, 2, 2)
plt.hist(NormalValues, bins=30, edgecolor='black')
plt.title('Histogram of Normal Distribution (mean=8, std=2)')
plt.xlabel('Waiting Time (minutes)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

#Part 3 - Statistical Analysis

```python
#for Uniform Distribution
print("For Uniform Distribution")
print("Mean:", np.mean(uniform_distributionValues))
print("Median:", np.median(uniform_distributionValues))
print("Standard Deviation:", np.std(uniform_distributionValues))
print("Maximum Wait Time:", np.max(uniform_distributionValues))


print("\n\n")

#For Normal Distribution
print("For Normal Distribution")
print("Mean:", np.mean(NormalValues))
print("Median:", np.median(NormalValues))
print("Standard Deviation:", np.std(NormalValues))
print("Maximum Wait Time:", np.max(NormalValues))
```
