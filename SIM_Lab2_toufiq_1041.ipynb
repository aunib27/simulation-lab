# Necessary Libraries

```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

# Matrix Functions

```python
A = np.array([[3, 5],
              [2, 6]])

print(A)



size_A = A.shape
print("Shape of A:", size_A)

det_A = np.linalg.det(A)
print("Determinant of A:", det_A)

rank_A = np.linalg.matrix_rank(A)
print("Rank of A:", rank_A)

eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues of A:", eigenvalues)

inv_A = np.linalg.inv(A)
print("Inverse of A:")
print(inv_A)
```

# Matrix Operations

```python
B = A.copy()
B[0, 0] += 1

print("Matrix B:")
print(B)

det_B = np.linalg.det(B)
print("Determinant of B:", det_B)

rank_B = np.linalg.matrix_rank(B)
print("Rank of B:", rank_B)

eigenvalues_B, eigenvectors_B = np.linalg.eig(B)
print("Eigenvalues of B:", eigenvalues_B)

inv_B = np.linalg.inv(B)

print("Inverse of B:")
print(inv_B)
```

7.Did the rank change? Explain.

The rank did not change. Both A and B have a rank of 2. Changing one element from 3 to 4 did not make the rows or columns dependent, so the rank remained 2.

8.How did the eigenvalues respond to the value change?

The eigenvalues changed after the matrix value was increased. The eigenvalues of A were 1 and 8, while the eigenvalues of B became approximately 1.6834 and 8.3166. This shows that changing even one matrix value can affect its eigenvalues.

9. Is B easier or harder to invert than A? Why?

B is easier to invert than A because its determinant is larger and farther from zero. The determinant of A is 8, while the determinant of B is 14. Since both determinants are non-zero, both matrices are invertible, but B has a larger determinant.
